```python
import os
import pandas as pd
from os.path import join
```

```python
src = os.path.join("misinfo-panel-austria", "data", "urls")
os.makedirs(src, exist_ok=True)
```

```python
urls = pd.read_csv(join(src, "shortened_urls_r2.csv.gz"),
                     compression="gzip")
len(urls), urls.url.nunique(), urls.domain.nunique()
```

```python
urls.domain.value_counts().head(15)
```

```python
last_batch = pd.read_csv(join(src, "unraveled_urls/", "unraveled_urls_23814.csv.gzip"), compression="gzip")

last_batch.status_code.value_counts()
```

```python
last_batch.tail(10)
```

```python
last_url = last_batch.url.iloc[-1]
last_url
```

```python
# whats the row number of this last url in urls?
last_row = urls[urls.url == last_url].index[0]
last_row
```

```python
# print the last 10 rows from that url
urls.iloc[last_row-10:last_row+1]
```

```python
# split urls into new and old depending on this row
new_urls = urls.iloc[last_row+1:]
new_urls.head(10)
```

```python
len(new_urls)
```

```python
new_urls.domain.value_counts().head(15)
```

```python
# remove youtu.be, fb.me and t.co
less_urls = new_urls[~new_urls.domain.isin([
    "youtu.be", "fb.me", "t.co", "instagr.am", "tmblr.co", "chang.it"])]
len(less_urls)
```

```python
less_urls.domain.value_counts().head(15)
```

```python
# save the new urls
less_urls.to_csv(join(src, 
"shortened_urls_reduc.csv.gzip"), 
                   compression="gzip", index=False)
```
